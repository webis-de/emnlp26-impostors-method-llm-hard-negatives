# Experiment: AV Quality per Similarity

We can assess semantic and syntactic similarity of paraphrases to the original text.
These similarity scores can used artificially bin paraphrases into groups based on their similarity.
Next, we can evaluate the quality of the AV model's predictions within each bin.
This will help us understand how the model performs across different levels of similarity.
We hypothesize that the model's performance may vary depending on how similar the paraphrases are to the original text:
Low similarity might lead to more FPs, while high similarity might lead to more FNs.
