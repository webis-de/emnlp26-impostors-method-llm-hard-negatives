# Reproduction of Koppel et al. (2014) 's paper

Koppel et al. (2014) proposed the Imposter approach alongside a supervised and an unsupervised baseline for authorship verification, which is implemented here in `impostor.py`, `impostor_supervised_baseline.py`, and `impostor_unsupervised_baseline.py` respectively. 
This notebook aims to reproduce the results of all three approaches on the same dataset (however, the input pairs may differ from the original paper).
The results from Koppel et. al. (2014) are most likely produced on the Blog dataset.

# Student Essay Dataset

```python
from datasets import load_from_disk
from genai_detection.config import CONFIG
import os
from pathlib import Path
import pandas as pd
```

```python
path2student_essays = Path(os.getcwd()).resolve().parent / CONFIG.PATH2STUDENT_ESSAYS
student_essay_train_dataset = load_from_disk(path2student_essays)["train"].to_pandas()
student_essay_test_dataset = load_from_disk(path2student_essays)["test"].to_pandas()
print(
    "Student Essay Dataset loaded successfully. Contains {} samples.".format(
        len(student_essay_train_dataset) + len(student_essay_test_dataset)
    )
)
display(student_essay_train_dataset.head())
print(student_essay_test_dataset["same"].value_counts())
print(student_essay_train_dataset["same"].value_counts())
```

# AV Models

```python
from genai_detection.detectors.impostor_supervised_baseline import (
    SupervisedImpostorBaseline,
)
from genai_detection.detectors.impostor_unsupervised_baseline import (
    UnSupervisedImpostorBaseline,
)
from genai_detection.detectors.impostor import ImpostorDetector

supervised_impostor_baseline = SupervisedImpostorBaseline()
print("Supervised Imposter Baseline initialized.")
unsupervised_impostor_baseline = UnSupervisedImpostorBaseline()
print("Unsupervised Imposter Baseline initialized.")
impostor = ImpostorDetector(
    impostor_technique="fixed",
    n_impostors=50,
    rounds=20,
    top_n=100000,
    path2imp=Path(os.getcwd()).resolve().parent / CONFIG.PATH2STUDENT_ESSAYS,
    upsample=False,
)
print("Imposter Detector initialized.")
```
