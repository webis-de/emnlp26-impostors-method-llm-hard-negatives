```python
import os
import sys
from pathlib import Path
import pandas as pd
import re
import unicodedata
from datasets import load_from_disk
from config import CONFIG

sys.path.append(os.path.abspath(".."))
```

```python
path = Path('../data/datasets/corpus-webis-authorship/koppel/')
```

```python
# df = pd.DataFrame(columns=['author', 'text'])


data = []
for author in path.iterdir():
    if author.is_dir():
        for file in author.iterdir():
            if file.is_file() and file.suffix == '.txt':
                with open(file, 'r', encoding='utf-8', errors='replace') as f:
                    content = f.read()
                    content = unicodedata.normalize("NFKC", content)
                    content = re.sub(r"[^\x00-\x7F]+", " ", content)
                    content = re.sub(r"\s+", " ", content)
                    content = content.strip().lower()
                    data.append({'author': author.name, 'text': content, 'n_chars': len(content)})
df = pd.DataFrame(data)
display(df)
```
