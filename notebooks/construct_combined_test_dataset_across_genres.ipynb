# Construction of Test dataset
When running experiments, I find the need to repeat the same experiment on different genres to assess whether the results are genre-specific or not. To facilitate this, I create a test dataset that includes samples from various genres in this notebook. 
On a more specific note, the dataset contains input pairs (i.e. text of unknown authorship and candidate text) from the **same genre** and their labels (i.e. whether the candidate text is the same as the text of unknown authorship or not).
Pairs originate from the Blog corpus, Gutenberg, Student Essays, and PAN20 Fanfiction corpus.

## Existing positive and negative label examples
